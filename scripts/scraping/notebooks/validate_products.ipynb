```python
# Đọc dữ liệu
import json
import pandas as pd

with open("../output/products_clean.json", "r", encoding="utf-8") as file:
    products = json.load(file)

df = pd.DataFrame(products)

print("Tổng sản phẩm:", len(products))
```

```python
# Kiểm tra số lượng giá trị rỗng theo từng trường
for column in df.columns:
    empty_count = 0

    for value in df[column]:
        if value == "" or value == []:
            empty_count += 1

    print(f"{column}: {empty_count}")
```

```python
# Xem colors nào xuất hiện:
df["colors"].value_counts()
```
